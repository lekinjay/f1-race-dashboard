```python
gas = clean_laps[clean_laps["Driver"] == "GAS"]
gas[["LapNumber", "TyreLife", "LapTimeSec"]].head(10)
```

```python
gas[["LapNumber", "TyreLife", "LapTimeSec"]].tail(10)
```

```python
import matplotlib.pyplot as plt

plt.plot(gas["TyreLife"], gas["LapTimeSec"], marker="o")
plt.xlabel("Tyre age (laps)")
plt.ylabel("Lap time (seconds)")
plt.title("GAS: lap time vs tyre age")
plt.show()
```

```python
gas = clean_laps[clean_laps["Driver"] == "GAS"]
bot = clean_laps[clean_laps["Driver"] == "BOT"]

plt.plot(gas["LapNumber"], gas["LapTimeSec"], label="GAS")
plt.plot(bot["LapNumber"], bot["LapTimeSec"], label="BOT")
plt.xlabel("Lap Number")
plt.ylabel("Lap time (seconds)")
plt.title("GAS VS BOT: lap time per lap")
plt.legend()
plt.show()
```

```python
gap = gas.merge(bot, on="LapNumber", suffixes=("_GAS", "_BOT"))
gap["Gap"] = gap["LapTimeSec_GAS"] - gap["LapTimeSec_BOT"]
gap[["LapNumber", "LapTimeSec_GAS", "LapTimeSec_BOT", "Gap"]].head(10)
```

```python
plt.plot(gap["LapNumber"], gap["Gap"], marker="o")
plt.axhline(0, color="black", linestyle="--")
plt.xlabel("LapNumber")
plt.ylabel("Gap: GAS minus BOT (seconds)")
plt.title("Lap-time gap: GAS VS BOT")
plt.show()
```

```python
gap["Gap"].mean()
```

```python
import numpy as np

late = gas[gas["LapNumber"] >= 48]
slope, intercept = np.polyfit(late["TyreLife"], late["LapTimeSec"], 1)
slope
```

```python
late_bot = bot[bot["LapNumber"] >= 48]
slope_bot, intercept_bot = np.polyfit(late_bot["TyreLife"], late_bot["LapTimeSec"], 1)
slope_bot
```

```python
plt.scatter(late["TyreLife"], late["LapTimeSec"], color="blue", label="GAS")
plt.plot(late["TyreLife"], slope * late["TyreLife"] + intercept, color="blue")

plt.scatter(late_bot["TyreLife"], late_bot["LapTimeSec"], color="orange", label="BOT")
plt.plot(late_bot["TyreLife"], slope_bot * late_bot["TyreLife"] + intercept_bot, color="orange")

plt.xlabel("Tyre age (laps)")
plt.ylabel("Lap time (seconds)")
plt.title("Tyre degradation: GAS VS BOT (lap 48 onward)")
plt.legend()
plt.show()
```

```python
med = clean_laps.groupby("LapNumber")["LapTimeSec"].median()

plt.plot(med.index, med.values, marker="o")
plt.xlabel("Lap Number")
plt.ylabel("Median lap time (seconds)")
plt.title("Track evolution: median lap time per lap")
plt.show()
```

```python
clean_laps2 = laps2[laps2["TrackStatus"] == "1"]
len(clean_laps2)
```

```python
clean_laps2 = clean_laps2.dropna(subset=["LapTime"])
len(clean_laps2)
```

```python
clean_laps2 = clean_laps2[clean_laps2["PitInTime"].isna() & clean_laps2["PitOutTime"].isna()]
len(clean_laps2)
```

```python
clean_laps2 = clean_laps2.copy()
clean_laps2["LapTimeSec"] = clean_laps2["LapTime"].dt.total_seconds()
clean_laps2[["Driver", "LapNumber", "LapTime", "LapTimeSec"]].head()
```

```python
two2 = clean_laps2[clean_laps2["Driver"].isin(["VER", "LEC"])]
two2.groupby(["Driver", "Stint", "Compound"]).size()
```

```python
ver2 = clean_laps2[(clean_laps2["Driver"] == "VER") & (clean_laps2["Stint"] == 2)]
lec2 = clean_laps2[(clean_laps2["Driver"] == "LEC") & (clean_laps2["Stint"] == 2)]

ver2[["LapNumber", "TyreLife", "LapTimeSec"]].head(10)
```

```python
lec2[["LapNumber", "TyreLife", "LapTimeSec"]].head(10)
```

```python
plt.scatter(ver2['TyreLife'], ver2['LapTimeSec'], color='blue', label='VER')
plt.scatter(lec2['TyreLife'], lec2['LapTimeSec'], color='red', label='LEC')

slope_ver2, intercept_ver2 = np.polyfit(ver2['TyreLife'], ver2['LapTimeSec'], 1)
slope_lec2, intercept_lec2 = np.polyfit(lec2['TyreLife'], lec2['LapTimeSec'], 1)

plt.plot(ver2['TyreLife'], slope_ver2 * ver2['TyreLife'] + intercept_ver2, color='blue')
plt.plot(lec2['TyreLife'], slope_lec2 * lec2['TyreLife'] + intercept_lec2, color='red')

plt.xlabel('Tyre age (laps)')
plt.ylabel('Lap time (seconds)')
plt.title('Tyre degradation: VER vs LEC (Stint 2, SOFT)')
plt.legend()
plt.show()
```

```python
print(f"VER degradation: {slope_ver2:.3f} sec per lap")
print(f"LEC degradation: {slope_lec2:.3f} sec per lap")
```
